# 3. Conditional Statements


```python
# Write a program that takes a number and prints "Positive", "Negative", or "Zero".
num = float(input("Enter a number: "))
print("Positive" if num > 0 else "Negative" if num < 0 else "Zero")
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
# Write a program that takes a student's marks (0-100) and prints their grade: 90+ = A, 75-89 = B, 60-74 = C, below 60 = F.
marks = float(input("Enter value: "))
print("A" if marks >= 90 else "B" if marks >=
      75 else "C" if marks >= 60 else "F")
```

```python
# Write a program that checks if a number is even or odd.
number = int(input('Enter any Number :'))
print("EVEN" if number % 2 == 0 else "ODD")
```

```python
# Take three numbers as input and print the largest one (using if-elif-else).
n1, n2, n3 = map(int, input('Enter any Number :'))
if n1 > n2:
    print("N1 is Greater", n1)
elif n2 > n3:
    print("N2 is Greater", n2)
else :
    print("N3 is Greater", n3)
```
