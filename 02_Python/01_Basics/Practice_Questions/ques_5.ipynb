```python
# Create a list of 5 course names (e.g., "SQL", "Python", "Power BI", etc.). Print the first and last item.
list = ['SQL', 'Python', 'Power BI']
print(f"List : {list} \n First_item : {list[0]} \n Last_item : {list[-1]}")
```

```python
# Add a new course to the end of the list, then remove one specific course.
list.append('StreamLit')
list.remove('Python')  # Remove the specific course

print(f"List: {list}")
```

```python
# Sort a list of numbers [45, 12, 89, 3, 67] in ascending and then descending order.
lst = [45, 12, 89, 3, 67]
lst.sort()
print(lst)
lst.sort(reverse=True)
print(lst)
```

```python
# Sample list of numbers
numbers = [45, 12, 89, 3, 22, 67, 99, 1]

# Check karo ki list khali toh nahi hai
if not numbers:
    print("List is empty !")
else:
    # 1. Variables initialize karo
    total_sum = 0
    # Shuruat mein pehle element ko hi max aur min maan lo
    max_val = numbers[0]
    min_val = numbers[0]

    # 2. Single loop chalakar teeno operations perform karo
    for num in numbers:
        # Sum calculate karo
        total_sum += num

        # Max find karo
        if num > max_val:
            max_val = num

        # Min find karo
        if num < min_val:
            min_val = num

    # 3. Results print karo
    print(f"List: {numbers}")
    print(f"Total Sum : {total_sum}")
    print(f"Maximum   : {max_val}")
    print(f"Minimum   : {min_val}")
```

```python
# Sample list of numbers
numbers = [45, 12, 89, 3, 22, 67, 99, 1]

# Built-in functions ka kamaal (Direct & Fast)
total_sum = sum(numbers)
max_val = max(numbers)
min_val = min(numbers)

# Results print karo
print(f"List: {numbers}")
print(f"Total Sum : {total_sum}")
print(f"Maximum   : {max_val}")
print(f"Minimum   : {min_val}")
```

```python
numbers = [10, 20, 30, 40, 50]

# Slicing with a negative step
reversed_list = numbers[::-1]

print("Original:", numbers)
print("Reversed:", reversed_list)
```
