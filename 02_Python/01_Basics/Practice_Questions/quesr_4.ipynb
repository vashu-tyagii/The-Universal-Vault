# 4. Loops


```python
# Print numbers from 1 to 20 using a for loop.
for i in range(1, 21):
    print(i)
```

```python
# Print only the even numbers from 1 to 50.
for i in range(1, 51, 2):
    print(i)
```

```python
# Calculate the sum of numbers from 1 to 100 using a loop.
total = 0

for i in range(1, 101):
    total += i

print(total)
```

```python
# Print the multiplication table of 7 (7x1 to 7x10).
i = int(input('Enter Table Number :'))
for j in range(1, 11):
    table = i * j
    print(f"{i} x {j} = {table}")
```

```python
# Using a while loop, keep asking the user for a number until they enter 0.
num = int(input('Enter any Number 1 to 10 :'))
while num != 0:
    num = int(input('Enter any Number 1 to 10 :'))
    print(num)
print('END !')
```
